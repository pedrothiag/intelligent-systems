# Algoritmos de Clusterização

## $k$-means

### Implementação com a `sklearn`

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
plt.rcParams["font.family"] = "Arial"
plt.rcParams["mathtext.fontset"] = "custom"
plt.rcParams['mathtext.rm'] = "Arial"
plt.rcParams['mathtext.it'] = "Arial:italic"
plt.rcParams['mathtext.bf'] = "Arial:bold"
plt.rcParams['figure.figsize'] = (7, 5)
```

```python
X, y = make_blobs(
    n_samples=400,
    centers=4,
    cluster_std=1,
    random_state=42,
)
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("Atributo 1")
plt.ylabel("Atributo 2")
plt.grid(alpha=0.3)
```

```python
modelo = KMeans(n_clusters=4, n_init=10, max_iter=100, tol=1e-3, random_state=42)
rotulos = modelo.fit_predict(X)
print(rotulos)
```

```python
print("\nCentróides:")
for i, c in enumerate(modelo.cluster_centers_):
    print(f"  m_{i+1} = ({c[0]:+.4f}, {c[1]:+.4f})")
```

```python
for c in range(4):
    mask = rotulos == c
    plt.scatter(X[mask, 0], X[mask, 1], s=35, alpha=0.7, label=f"Cluster {c+1}")
plt.scatter(modelo.cluster_centers_[:, 0], modelo.cluster_centers_[:, 1],
            marker="X", s=220, c="black", edgecolors="white", linewidths=2,
            label="Centróides", zorder=5)
plt.title("Clusters do k-means")
plt.xlabel("Atributo 1")
plt.ylabel("Atributo 2")
plt.legend()
plt.grid(alpha=0.3)
```

### Implementação Manual

```python
"""
n_clusters : Número de clusters (k).
max_iter : Máximo de iterações por inicialização.
tol : Tolerância de convergência (deslocamento máximo dos centróides).
"""
class myKMeans:
    def __init__(self, n_clusters=8, n_init=10, max_iter=300,
                 tol=1e-4, random_state=None):
        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.tol = tol
        self.random_state = random_state
        self.cluster_centers_ = None   # shape (n_clusters, n_features)
        self.labels_ = None            # shape (n_samples,)
        self.n_init = n_init
        self.inertia_ = None

    """Seleciona k objetos aleatórios do conjunto de dados como centróides"""
    def _init_centroids(self, X, rng):
        idx = rng.choice(len(X), size=self.n_clusters, replace=False)
        return X[idx].copy()

    """Atribui cada ponto ao centróide mais próximo"""
    def _assign(self, X, centroids):
        sq_dists = np.zeros((len(X), self.n_clusters))
        for j in range(self.n_clusters):
            diff = X - centroids[j]
            sq_dists[:, j] = np.sum(diff ** 2, axis=1)
        return np.argmin(sq_dists, axis=1)

    """Recalcula centróides como média dos pontos atribuídos"""
    def _update(self, X, labels, centroids, rng):
        new = np.zeros_like(centroids)
        for j in range(self.n_clusters):
            mask = labels == j
            new[j] = X[mask].mean(axis=0) if mask.any() else X[rng.integers(0, len(X))]
        return new

    """Função para Calcular a inércia"""
    def _inertia(self, X, labels, centroids):
        return sum(
            np.sum((X[labels == j] - centroids[j]) ** 2)
            for j in range(self.n_clusters)
            if (labels == j).any()
        )

    """Função para realizar apenas uma execução"""
    def _run_once(self, X, rng):
        centroids = self._init_centroids(X, rng)
        labels = self._assign(X, centroids)
        for i in range(self.max_iter):
            new_centroids = self._update(X, labels, centroids, rng)
            labels = self._assign(X, new_centroids)
            shift = np.max(np.linalg.norm(new_centroids - centroids, axis=1))
            centroids = new_centroids
            if shift < self.tol:
                return centroids, labels, self._inertia(X, labels, centroids), i + 1
        return centroids, labels, self._inertia(X, labels, centroids), self.max_iter

    """Função para realizar o ajuste do modelo - realiza vários testes e seleciona o com menor inércia"""
    def fit(self, X):
        X = np.asarray(X, dtype=float)
        rng = np.random.default_rng(self.random_state)

        best_inertia = np.inf
        for _ in range(self.n_init):
            centroids, labels, inertia, n_iter = self._run_once(X, rng)
            if inertia < best_inertia:
                best_inertia = inertia
                self.cluster_centers_ = centroids
                self.labels_ = labels
                self.inertia_ = inertia

        return self

    """Ajusta o modelo e retorna os rótulos"""
    def fit_predict(self, X):
        return self.fit(X).labels_

    """Função para realizar a predição"""
    def predict(self, X):
        X = np.asarray(X, dtype=float)
        return self._assign(X, self.cluster_centers_)
```

```python
X, y = make_blobs(
    n_samples=400,
    centers=4,
    cluster_std=1,
    random_state=42,
)
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("Atributo 1")
plt.ylabel("Atributo 2")
plt.grid(alpha=0.3)
```

```python
modelo = myKMeans(n_clusters=4, n_init=10, random_state=42)
rotulos = modelo.fit_predict(X)
print(rotulos)
```
